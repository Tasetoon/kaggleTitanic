```python
import pandas as pd
from sklearn.model_selection import train_test_split
from eda.preprocessing import prepare_data

train = prepare_data('data/train.csv')
```

```python
train.head()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report

features = ['Pclass', 'Sex', 'Age', 'Fare', 'Relatives', 'Embarked_Q', 'Embarked_S']
X = train[features]

Y = train['Survived']

model = DecisionTreeClassifier(max_depth=6, random_state=42)

X_train, X_val, Y_train, Y_val = train_test_split(X, Y, test_size=0.2, random_state=42)
model.fit(X_train, Y_train)

print(classification_report(Y_val, model.predict(X_val)))
```

```python
test = prepare_data('data/test.csv')

test_predict = model.predict(test[features])

prediction = pd.DataFrame()

ids = pd.read_csv('data/test.csv')['PassengerId']

prediction['PassengerId'] = ids
prediction['Survived'] = test_predict
prediction.to_csv('prediction_tree.csv', index=False)
```
